## Merging , Joining , Concatenating

### Merging

```python
import pandas as pd

askari_df = pd.DataFrame({
    'Name': ['Ali', 'Danial', 'Rezvan', 'Moosa'],
    'GPA': [20, 19, 15, 20],
    'Class': ['A', 'B', 'C', 'D']
})

razavi_df = pd.DataFrame({
    'Absence': [2, 0, 5, 0],
    'Food': [True, False, False, False],
    'Class': ['A', 'B', 'C', 'D']
})

merged = pd.merge(askari_df, razavi_df, on='Class')
print(merged)
```

### Joining

```python
askari_df = pd.DataFrame({
    'Name': ['Ali', 'Danial', 'Rezvan', 'Moosa'],
    'GPA': [20, 19, 15, 20],
}, index = ['A', 'B', 'C', 'D'])

razavi_df = pd.DataFrame({
    'Absence': [2, 0, 5, 0],
    'Food': [True, False, False, False],
}, index = ['A', 'B', 'C', 'D'])

joined_df = askari_df.join(razavi_df)
print(joined_df)
```

### Concatenating

#### With Same Columns

```python
import pandas as pd

askari_df = pd.DataFrame({
    'Name': ['Ali', 'Danial', 'Rezvan', 'Moosa'],
    'GPA': [20, 19, 15, 20],
})

razavi_df = pd.DataFrame({
    'Name': ['Reza', 'Hadi', 'Milad', 'Sana'],
    'GPA': [12, 11, 14, 12],
})

concatenated = pd.concat([askari_df, razavi_df], axis=0, ignore_index=True)
print(concatenated)
```

#### With Different Columns

```python
askari_df = pd.DataFrame({
    'Name': ['Ali', 'Danial', 'Rezvan', 'Moosa'],
    'GPA': [20, 19, 15, 20],
})

razavi_df = pd.DataFrame({
    'Absence': [2, 0, 5, 0],
    'Food': [True, False, False, False],
})

concatenated = pd.concat([askari_df, razavi_df], axis=1)
print(concatenated)
```
